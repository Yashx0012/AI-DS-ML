MatplotLib Tutorial

Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt, or GTK+.

Some of the major Pros of Matplotlib are:

1.   Generally easy to get started for simple plots
2.   Support for custom labels and texts
3.   Great control of every element in a figure
4.   High-quality output in many formats
5.   Very customizable in general

**SUBPLOT**

subplot is the matplotlib command that creates one specific panel inside the figure that is laid out of number of rows by numbers of column grid

for example: plt.subplot(2,2,2)

| Argument   | Meaning                                                                              |
| ---------- | ------------------------------------------------------------------------------------ |
| 2 (first)  | Number of rows in the grid                                                           |
| 2 (second) | Number of columns in the grid                                                        |
| 2 (third)  | The index (position) of the subplot (starts from 1, goes left → right, top → bottom) |

subplot positions:
subplot(2,2,1) → top-left
subplot(2,2,2) → top-right
subplot(2,2,3) → bottom-left
subplot(2,2,4) → bottom-right


```python
#compute x and y coordinates for the points on sine and cosine curves 
x=np.arange(0, 5*np.pi, 0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)
# Set up a subplot grid that has height 2 and width 1, 
# and set the first such subplot as active. 
plt.subplot(2,1,1)
plt.plot(x,y_sin, 'r--')
plt.title('sine wave form')
plt.subplot(2,1,2)
plt.plot(x,y_cos, 'g--')
plt.title('cosine wave form')
plt.tight_layout()
plt.show()

```

Box Plot using Matplotlib

* data = [np.random.normal(0, std, 100) for std in range(1, 4)]

**What this does:**

- Creates a list of datasets

- Each dataset has 100 random values

- Values follow a normal distribution

Breakdown:

1. np.random.normal(mean, std_dev, size)

2. mean = 0

3. std = 1, 2, 3 (because of range(1,4))

4. size = 100

[
  dataset1 (std=1),
  dataset2 (std=2),
  dataset3 (std=3)
]

- each data will have different spreead which means variation

**plt.boxplot(data, vert=True, patch_artist=False)**

   vert=True

- Makes the boxplots vertical

   If:

    vert=False

- then boxplots become horizontal

**patch_artist=False**

- Controls color filling

 If:

 False → boxes are empty (only outline)

 True → boxes are filled with color 

```python
data=[np.random.normal(1, std, 100)
      for std in range(1,4)]
plt.boxplot(data, vert=True, patch_artist=False)
```

**BAR PLOT**

```python
#data to plot
labels=['python', 'C++', 'C', 'Java']
sizes=[215, 130, 245, 210]
colors=['gold','lightcoral','lightblue', 'lightgreen']
explode=[0.1,0.2,0.3,0]

#PLOTTING
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
        shadow=True)
plt.axis('equal')
plt.show()
```

These are the values (data) for each category.

1. Python → 215
2. C++ → 130
3. Ruby → 245
4. Java → 210

The pie chart will calculate percentages based on these values.

-------------------------------------------------------------------------------------------------------------------------------------------

autopct='%1.1f%%'

- This shows percentage values on the chart

1. %1.1f → float with 1 decimal place
2. %% → shows % symbol

* Example output:

30.2%

-------------------------------------------------------------------------------------------------------------------------------------------

shadow=False

- No shadow effect under the pie

If True, adds a shadow for 3D effect

-------------------------------------------------------------------------------------------------------------------------------------------

plt.axis('equal')

- Ensures the pie chart is a perfect circle

Without this:

- Pie may look stretched (oval shape)

-------------------------------------------------------------------------------------------------------------------------------------------

**EXPLODE**

- explode controls how much each slice moves outward (separates) from the center of the pie chart.

| Index | Label  | Value | Meaning                 |
| ----- | ------ | ----- | ----------------------- |
| 0     | Python | 0.4   | Move this slice outward |
| 1     | C++    | 0     | No movement             |
| 2     | Ruby   | 0     | No movement             |
| 3     | Java   | 0     | No movement             |

It is the distance from the center (as a fraction of the radius)

1. 0 → no movement
2. 0.1 → small separation
3. 0.4 → noticeable separation
4. 1 → very far away
