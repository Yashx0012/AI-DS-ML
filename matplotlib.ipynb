MatplotLib Tutorial

Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt, or GTK+.

Some of the major Pros of Matplotlib are:

1.   Generally easy to get started for simple plots
2.   Support for custom labels and texts
3.   Great control of every element in a figure
4.   High-quality output in many formats
5.   Very customizable in general

 %matplotlib inline means “Show graph here only (inside notebook), not in new window.”

```python
import matplotlib.pyplot as plt
%matplotlib  inline 
```

```python
import numpy as np
x=np.arange(0,10)
y=np.arange(10,20)
a=np.arange(40,50)
b=np.arange(50,60)
```

```python
# plotting with matplotlib
#plt scatter it will be scattering the values of x and y 
plt.scatter(x,y,c='g')
# c= g means that color and green 
#since i am using this %matplotlib inline show its doesnt required to write a code for show the graph that pl.show()
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('2d graph')
plt.savefig('test.png')
```

```python
#plt plot means basically u want a straight line in the  graph
y=x*x
plt.plot(x,y)
```

```python
plt.plot(x,y, 'r--')
```

```python
plt.plot(x,y,'r*-')
```

```python
plt.plot(x,y,'ro')
```

```python
plt.plot(x, y, 'r*', linestyle='dashed', linewidth=1, markersize=12)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('2d diagram')
```

```python
#creating subplot which means creating multiple graph
plt.subplot(2,2,1)
plt.plot(x,y,c='r')
plt.subplot(2,2,2)
plt.plot(x,y,'b-*')
plt.subplot(2,2,3)
plt.plot(x,y,'g--')
```

```python
x1=np.arange(1,11)
y1= 3*x1+5
plt.title('matplotlib demo')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.plot(x1,y1)
plt.show()
```

```python
# computing x and y coordinates in the sine wave form 
#np.arange(start, stop, step) creates a sequence of numbers from 0 up to (but not including) 4π (≈ 12.566), increasing by 0.1 each step.
x=np.arange(0, 4*np.pi, 0.1)
y=np.sin(x)
plt.title('sine wave form')
plt.plot(x,y)
plt.show()
```
