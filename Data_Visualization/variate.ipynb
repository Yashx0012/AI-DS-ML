````python
df_setosa=df.loc[df['species']=='setosa']
df_virginica=df.loc[df['species']=='virginica']
df_versicolor=df.loc[df['species']=='versicolor']
````

All points lie on the horizontal line y = 0.

This technique is often used in 1D data visualization (rug plots) to show the distribution of sepal lengths along the x-axis without considering a second variable. 

The 'o' is a marker style.
- 'o' = draw a circle at each data point.
- Without 'o', Matplotlib connects the points with lines by default.

````python
plt.plot(df_setosa['sepal_length'],np.zeros_like(df_setosa['sepal_length']),'o')
plt.plot(df_virginica['sepal_length'],np.zeros_like(df_virginica['sepal_length']),'o')
plt.plot(df_versicolor['sepal_length'],np.zeros_like(df_versicolor['sepal_length']),'o')
plt.xlabel('sepal_length')
plt.show()
````

## Bivariate Analysis

### Scatter Plot using Seaborn FacetGrid

```python
sns.FacetGrid(df, hue="species", height=5)\
    .map(plt.scatter, "sepal_length", "sepal_width")\
    .add_legend()
```

### Explanation

#### 1. `sns.FacetGrid(df, hue="species", height=5)`
- Creates a plotting grid using the DataFrame `df`.
- `hue="species"` assigns different colors to different flower species.
- `height=5` sets the height of the plot to 5 inches.

#### 2. `.map(plt.scatter, "sepal_length", "sepal_width")`
- Maps a scatter plot onto the grid.
- `sepal_length` is plotted on the X-axis.
- `sepal_width` is plotted on the Y-axis.
- Each point represents a flower sample.

#### 3. `.add_legend()`
- Adds a legend to identify the colors corresponding to each species.

### Purpose
This visualization helps analyze the relationship between **Sepal Length** and **Sepal Width** for different Iris flower species. Different colors make it easier to observe patterns and separability among species.

### Observation
- Setosa samples form a distinct cluster.
- Versicolor and Virginica show some overlap.
- These features can be useful for classification tasks in Machine Learning.
