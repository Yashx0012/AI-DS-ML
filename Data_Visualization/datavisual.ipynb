```python
import seaborn as sns
```

```python
df=sns.load_dataset('Data/iris')
df.head(6)
```

```python
df.dtypes
```

**CORRELATION WITH HEATMAP**

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df. corr(numeric_only=True))
```

**UNIVARIATE ANALYSIS**

```python
sns.jointplot(x='sepal_length', y='sepal_width', data=df, kind='hex')
```

```python
sns.jointplot(x='petal_length', y='petal_width', data=df, kind='hex')
```

```python
sns.jointplot(x='sepal_length', y='petal_width', data=df, kind='reg')
```

```python
sns.jointplot(x='petal_length', y='sepal_width', data=df, kind='reg')
```
