****be careful when the copying the array

```python
a=np.array([1,2,3])
b=a
b[0]=100
print(a)
#b=a creates a reference, not a copy, so modifying b also changes a.
```

```python
a=np.array([1,2,3])
b=a.copy()
b[0]=100
print(a)#Then changes in b will NOT affect a.
```
