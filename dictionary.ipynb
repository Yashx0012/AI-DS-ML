```python
#nested dictionary 
car_model1={'mercedes':1960}
car_model2={'bmw':1970}
car_model3={'audi':1980}
car_type={'car1':car_model1, 'car2':car_model2, 'car3':car_model3}
print(car_type)
```

```python
#ACCESSING THE ITEM IN THE DICTIONARY
print(car_type["car1"])
```
