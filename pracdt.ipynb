```python
import pandas as pd
import numpy as np
```

```python
data = {
"Name": ["Alice", "Bob", "Alice", "Charlie", "Bob", "David"],
"age": [25, 30, 25, 35, 30, 40],
"Salary": [50000, 60000, 50000, 70000, 60000, 80000],
"City": ["NY", "LA", "NY", "SF", "LA", "NY"]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
```

```python
df['normal_age']=(df['age']-df['age'].min())/(df['age'].max()-df['age'].min())
```

```python
df['City']=df['City'].replace("NY", "NewYok")
```

```python
df['City']=df['City'].replace({'NewYok':'NY', "LA":'Losangels'})
```

```python
df['age']=df['age'].apply(lambda x : "senior" if x>30 else 'junior')
```

```python
data = {
"A": [1, 2, np.nan, 4, 5],
"B": [np.nan, 2, 3, 4, np.nan],
"C": ["cat", "dog", np.nan, "rabbit", "cat"]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
```

```python
df['A']=df['A'].fillna(df["A"].mean())
df["B"]=df["B"].fillna(df['B'].median())
```

```python
df['A']=df['A'].interpolate(method='linear')
df["A"]
```

```python
data= {"value":[10, 12,14,15,100, 16, 18, 19]}
df = pd.DataFrame(data)
df['value'].mean()
df['value'].std()
df['z_score']=(df['value']-df['value'].mean())/df['value'].std()
filtered_df=df[np.abs(df['value'])<2]
```

```python
data = {"value": [10, 12, 14, 15, 100, 16, 18, 19]}
df = pd.DataFrame(data)
print(df)
q1=df["value"].quantile(0.25)
q3=df["value"].quantile(0.75)
IQR=q3-q1
lower_bound=q1-1.5*IQR
upper_bound=q3+1.5*IQR
filtered_df=df[(df["value"]>=lower_bound) & (df["value"]<=upper_bound)]
```
