```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
data={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,np.nan],
    'c':["cat", "dog", np.nan, "rabbit", "cat"]

}
df=pd.DataFrame(data)
print('original datset:')
print(df)
```

Dropping Missing Values

```python
df_dopped_rows=df.dropna()
print("\n row with missing values will be removed")
print(df_dopped_rows)
df_dopped_columns=df.dropna(axis=1)# every element that contain nan values will be removed 
print('\n coloumns with missing values removed:')
print(df_dopped_columns)
```

✅ What is Imputation?

👉 Imputation = filling NaN values with meaningful values

Instead of:

dropna()


(which removes data)

we do:

fillna()


(which keeps data and replaces NaN)

This is very important in Data Science because:

✔ you don’t lose rows
✔ models work better
✔ dataset remains complete
-------------------------------------------------------------------------------

Missing numerical values are replaced with the average (mean) of the column.
Missing categorical values are replaced with the most frequent value (mode).
Missing numerical values are replaced with the middle value (median).

```python
data={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,np.nan],
    'c':["cat", "dog", np.nan, "rabbit", "cat"]

}
df=pd.DataFrame(data)
print('original datset:')
print(df)
```

```python
df['a']=df['a'].fillna(df['a'].mean())
df['b']=df['b'].fillna(df['b'].median())
df['c']=df['c'].fillna(df['c'].mode()[0])
print('\n after mean mode median imputation')
print(df)
```

forward fill and backward fill 
Fill NaN using the previous (above) value
Fill NaN using the next (below) value

```python
data={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,np.nan],
    'c':["cat", "dog", np.nan, "rabbit", "cat"]

}
df=pd.DataFrame(data)
print('original datset:')
print(df)
#forward fill
df_ffill=df.ffill()
print('\nforwar fill ')
print(df_ffill)
#backward fill
df_bfill=df.bfill()
print('\nbackward fill')
print(df_bfill)
```

interpolation
Interpolation means:

Fill missing values using a mathematical estimation based on surrounding values.

It does not use mean or mode.
It calculates a value between two known values.
Linear interpolation works only on NUMBERS, not text.

```python
data={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,np.nan],
    'c':["cat", "dog", np.nan, "rabbit", "cat"]

}
df=pd.DataFrame(data)
print('original datset:')
print(df)
#interpolate missing values
df['a']=df['a'].interpolate(method='linear')
print('\n after linear interpolation ')
print(df)
```

(2 + 4) / 2 = 3.0

knn imputation
neighbour(k)
| Dataset size | K    |
| ------------ | ---- |
| Very small   | 2–3  |
| Medium       | 3–7  |
| Large        | 5–15 |
Although any value of K can be chosen, very small K causes noisy results and very large K makes imputation similar to mean. Therefore K must be selected carefully based on dataset size.

📊 Dataset size (rough guideline)
🔹 Small dataset

👉 < 100 rows

Example:

Student marks of one class (30–50 students)

Your current toy dataframe (5 rows)

Typical KNN choice: K = 2 or 3

🔹 Medium dataset

👉 100 to ~10,000 rows

Examples:

Sales data for 1–2 years (500–5000 records)

Employee database of a company (300–2000 employees)

Survey responses (1000 people)

Sensor readings for a few months

Typical KNN choice:

K = 3 to 7


Why?
Because you now have enough nearby points to average without becoming too noisy or too global.

🔹 Large dataset

👉 > 10,000 rows

Examples:

E-commerce logs

Web click data

Large ML datasets

Typical KNN choice:

K = 5 to 15


(also KNN becomes slow here)

4️⃣ Apply KNN
knn_imputer.fit_transform(...)

What happens internally:
a) Measure distance between rows

(using Euclidean distance)

b) Find 2 nearest rows (neighbors)
c) Take the average of those neighbors for the missing value

Example (simple):

If value is missing in row X:

Nearest values are:

2 and 4


KNN fills:

(2 + 4) / 2 = 3



```python
data={
    'a':[1,2,np.nan,4,5],
    'b':[np.nan,2,3,4,np.nan],
    'c':["cat", "dog", np.nan, "rabbit", "cat"]

}
df=pd.DataFrame(data)
print('original datset:')
print(df)

knn_imputer=KNNImputer(n_neighbors=2)
df_knn=pd.DataFrame(knn_imputer.fit_transform(df.iloc[:, :2]), columns=['a','b'])
print('\n after knn imputation ')
print(df_knn)
#iloc = integer-location based indexing
# It is used to select rows and columns by their numeric position, not by name.
```


MICE = Multiple Imputation by Chained Equations

In simple words:

MICE fills missing values by building prediction models between columns and repeating this process multiple times.

```python
iter_imputer=IterativeImputer(max_iter=10, random_state=0)
df_mice=pd.DataFrame(iter_imputer.fit_transform(df.iloc[:, :2]), columns=['a','b'])
print('\n after mice imputaion')
print(df_mice)
df_mice = df_mice.round()
print(df_mice)
```

Placeholder Value Assignment
t means:

Replace missing values with fixed dummy values (placeholders).

Instead of calculating mean / KNN / MICE, you manually tell Pandas:

👉 “Whenever you see NaN, replace it with THIS.”

```python
df_placeholder=df.fillna(value={'a':0,'b':0,'c':'unknown'})
print("\n after placeholder")
print(df_placeholder)
```
