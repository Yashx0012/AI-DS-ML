**** apply() applies a function to every value of a Series or to each row/column of a DataFrame.

applymap() applies a function to EVERY single cell of the DataFrame.

map() replaces values using a dictionary or function.

apply()

Used to apply a function to each value of a column or row.

applymap()

Used to apply a function to every element of a DataFrame.

map()

Used on Series to replace or transform values.****


???Lambda is an anonymous one-line function in Python used for small, simple operations without defining a full function. like def function

lambda x → take each salary value

x + 1000 → add 1000

apply() → apply to every row???



==apply() works on a column or row, while applymap() works on every element of the DataFrame.==

```python
df['salary']=df['salary'].apply(lambda x : x+1000)# APPLY->applies a function to every value of a Series or to each row/column of a DataFrame.
df
```

```python
df[['marks','age','salary']] = df[['marks','age','salary']].apply(lambda col: col*2)
df
```

rank() assigns ranking numbers to values of a column. By default, lowest value gets rank 1.
tanish → 1.0  (lowest salary)
pranjal → 2.0
shruti → 3.0
yash → 4.0
tejas → 5.0
lavanya → 6.0 (highest salary)


```python
import pandas as pd
import numpy as np

data = {
    'name': ['yash','tejas','shruti','tanish','lavanya','pranjal'],
    'marks': [92, 88, 40, 45, 86, 76],
    'city': ['itarsi', 'dabra', 'betul', 'katni','gwalior', 'morena'],
    'age': [21, 24, 22, 19, 18, 20],
    'salary': [2500, 3000, 2200, 1500, 4000, 1700]
}

df = pd.DataFrame(data)
df
```

```python
df['rank']=df['salary'].rank()
df
```

| Name    | Salary | Rank |
| ------- | ------ | ---- |
| lavanya | 4000   | 1    |
| tejas   | 3000   | 2    |
| yash    | 2500   | 3    |
| shruti  | 2200   | 4    |
| pranjal | 1700   | 5    |
| tanish  | 1500   | 6    |


```python
df['rank']=df['salary'].rank(ascending=False)
df
```
